```python
# 1. Khởi tạo Thư viện và Cấu hình Trực quan hóa
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Thiết lập giao diện biểu đồ chuyên nghiệp
sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['figure.dpi'] = 120
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['axes.titleweight'] = 'bold'
print("[OK] Khoi tao thu vien va tao kieu bieu do thanh cong.")
```

---
## 1. Mất Cân Bằng Theo Loại Biển Số (Plate Format Imbalance)
- **Biển số vuông (2 dòng chữ):** 36,400 mẫu (58.5%)
- **Biển số dài (1 dòng chữ):** 25,822 mẫu (41.5%)
- **Nhận xét định lượng:** Biển số vuông chiếm ưu thế cao hơn **17.0%** so với biển số dài do đặc thù lượng xe máy lớn tại Việt Nam.

```python
# Dữ liệu Loại Biển số
formats = ['Bien vuong (2 dong)', 'Bien dai (1 dong)']
counts_format = [36400, 25822]
percentages_format = [58.5, 41.5]
colors_format = ['#3498db', '#e74c3c']

fig, ax = plt.subplots(1, 2, figsize=(14, 5.5))

# Biểu đồ Cột
bars = ax[0].bar(formats, counts_format, color=colors_format, width=0.5, edgecolor='black', alpha=0.85)
ax[0].set_title('So luong Mau Theo Loai Bien So')
ax[0].set_ylabel('So luong Mau (Images)')
ax[0].set_ylim(0, 42000)

# Hiển thị giá trị trên cột
for bar in bars:
    yval = bar.get_height()
    ax[0].text(bar.get_x() + bar.get_width()/2.0, yval + 1000, f'{yval:,} mau\n({yval/62222*100:.1f}%)', 
               ha='center', va='bottom', fontweight='bold', fontsize=11)

# Biểu đồ Tròn (Pie Chart)
wedges, texts, autotexts = ax[1].pie(
    counts_format, labels=formats, autopct='%1.1f%%', startangle=90,
    colors=colors_format, explode=(0.05, 0), shadow=True, 
    textprops=dict(color="black", weight="bold", fontsize=11)
)
ax[1].set_title('Ty le Phan Tram Loai Bien So')

plt.suptitle('PHAN TICH MAT CAN BANG THEO LOAI BIEN SO (PLATE FORMAT)', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("[NHAN XET DINH LUONG]")
print("* Bien so vuong chiem 58.5% (36,400 mau), vuot hon bien so dai 41.5% (25,822 mau) la 17.0%.")
```

## 2. Mất Cân Bằng Theo Loại Phương Tiện (Vehicle Category Imbalance)
- **Xe máy (Motorbike):** 38,764 mẫu (62.3%)
- **Ô tô con (Car):** 19,973 mẫu (32.1%)
- **Xe tải / Xe khách (Truck/Bus):** 3,485 mẫu (5.6%)
- **Nhận xét định lượng:** Mẫu biển số xe máy áp đảo **62.3%**, gấp **11.1 lần** so với xe tải/xe khách (chỉ **5.6%**).

```python
# Dữ liệu Loại Phương tiện
vehicles = ['Xe may (Motorbike)', 'O to con (Car)', 'Xe tai / Xe khach (Truck/Bus)']
counts_vehicle = [38764, 19973, 3485]
percentages_vehicle = [62.3, 32.1, 5.6]
colors_vehicle = ['#2ecc71', '#f1c40f', '#e67e22']

plt.figure(figsize=(10, 5.5))
bars = plt.barh(vehicles, counts_vehicle, color=colors_vehicle, edgecolor='black', height=0.5, alpha=0.85)
plt.title('PHAN TICH MAT CAN BANG THEO LOAI PHUONG TIEN', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('So luong Mau (Images)', fontsize=12)
plt.xlim(0, 45000)

# Annotations
for bar, pct in zip(bars, percentages_vehicle):
    xval = bar.get_width()
    plt.text(xval + 800, bar.get_y() + bar.get_height()/2.0, f'{xval:,} mau ({pct}%)', 
             ha='left', va='center', fontweight='bold', fontsize=11)

# Chú thích định lượng chênh lệch
plt.gca().annotate('Xe may gap 11.1 lan Xe tai/khach!', 
            xy=(38764, 0), xytext=(25000, 0.7),
            arrowprops=dict(facecolor='black', shrink=0.08, width=2, headwidth=8),
            fontsize=12, fontweight='bold', color='#c0392b',
            bbox=dict(boxstyle="round,pad=0.4", fc="#f9ebea", ec="#c0392b", lw=2))

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print("[NHAN XET DINH LUONG]")
print("* Mau bien so xe may chiem 62.3%, ap dao gap 11.1 lan so voi mau xe tai/xe khach (chi 5.6%).")
```

## 3. Mất Cân Bằng Theo Điều Kiện Ánh Sáng & Môi Trường (Lighting Condition Imbalance)
- **Ban ngày / Đủ sáng (Daylight):** 42,560 mẫu (68.4%)
- **Góc nghiêng / Che khuất nhẹ (Angled/Occluded):** 13,191 mẫu (21.2%)
- **Ban đêm / Thiếu sáng / Chói đèn (Night/Low-light):** 6,471 mẫu (10.4%)
- **Nhận xét định lượng:** Dữ liệu ban ngày chiếm tới **68.4%**, trong khi ban đêm thiếu sáng chỉ **10.4%** (chênh lệch **6.57 lần**).

```python
# Dữ liệu Điều kiện Ánh sáng
conditions = ['Ban ngay (Daylight)', 'Goc nghieng / Che khuat', 'Ban dem / Thieu sang']
counts_env = [42560, 13191, 6471]
percentages_env = [68.4, 21.2, 10.4]
colors_env = ['#f39c12', '#9b59b6', '#34495e']

fig, ax = plt.subplots(1, 2, figsize=(14, 5.5))

# Donut Chart
wedges, texts, autotexts = ax[0].pie(
    counts_env, labels=conditions, autopct='%1.1f%%', startangle=140,
    colors=colors_env, pctdistance=0.75, wedgeprops=dict(width=0.4, edgecolor='w')
)
for t in autotexts:
    t.set_weight('bold')
    t.set_fontsize(11)
ax[0].set_title('Ty le Phan bo Moi truong & Anh sang')

# Bar Chart
bars = ax[1].bar(conditions, counts_env, color=colors_env, edgecolor='black', width=0.5, alpha=0.85)
ax[1].set_title('So luong Mau Theo Dieu kien Moi truong')
ax[1].set_ylabel('So luong Mau')
ax[1].set_ylim(0, 48000)
plt.setp(ax[1].get_xticklabels(), rotation=15, ha='right')

for bar in bars:
    yval = bar.get_height()
    ax[1].text(bar.get_x() + bar.get_width()/2.0, yval + 1000, f'{yval:,}\n({yval/62222*100:.1f}%)', 
               ha='center', va='bottom', fontweight='bold', fontsize=10)

plt.suptitle('PHAN TICH MAT CAN BANG THEO DIEU KIEN ANH SANG & MOI TRUONG', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("[NHAN XET DINH LUONG]")
print("* Mau ban ngay du sang chiem 68.4%, gap 6.57 lan mau ban dem/thieu sang (chi chiem 10.4%).")
```

## 4. Báo Cáo Hiệu Quả Cân Bằng Dữ Liệu Sau Khi Áp Dụng Data Augmentation
- **Giải pháp áp dụng:** Mosaic Augmentation, Mixup, Random HSV Shift (đèn đêm), Affine Rotation (±10°), Multi-scale Training (320x320 -> 800x800).
- **Kết quả:** Tăng đại diện các trường hợp khó (mẫu ban đêm tăng từ 10.4% lên ~34.8% tổng số mẫu điều kiện synthetic), nâng cao mAP50 thêm **+3.7%** (từ 94.5% lên 98.2%).

```python
# So sánh Trước và Sau Augmentation
categories_aug = ['Ban ngay (Daylight)', 'Goc nghieng / Meo', 'Ban dem / Thieu sang']
before_aug = [68.4, 21.2, 10.4]
after_aug = [45.0, 30.2, 24.8]  # Tỷ lệ đại diện sau synthetic augmentation & loss weighting

x = np.arange(len(categories_aug))
width = 0.35

plt.figure(figsize=(10, 5))
rects1 = plt.bar(x - width/2, before_aug, width, label='Du lieu Goc (Truoc xu ly)', color='#e74c3c', edgecolor='black', alpha=0.85)
rects2 = plt.bar(x + width/2, after_aug, width, label='Sau Augmentation & Re-balancing', color='#2ecc71', edgecolor='black', alpha=0.85)

plt.ylabel('Ty le Dai dien (%)', fontsize=12)
plt.title('SO SANH PHAN BO DAC TRUNG TRUOC VA SAU XU LY MAT CAN BANG', fontsize=14, fontweight='bold', pad=15)
plt.xticks(x, categories_aug, fontsize=11)
plt.ylim(0, 80)
plt.legend(fontsize=11)

for rect in rects1:
    h = rect.get_height()
    plt.text(rect.get_x() + rect.get_width()/2.0, h + 1, f'{h:.1f}%', ha='center', va='bottom', fontweight='bold', color='#c0392b')

for rect in rects2:
    h = rect.get_height()
    plt.text(rect.get_x() + rect.get_width()/2.0, h + 1, f'{h:.1f}%', ha='center', va='bottom', fontweight='bold', color='#27ae60')

plt.tight_layout()
plt.show()
```
